# Model Alignment of Gemma

Model alignment is the process of automatically updating a prompt based on developer-provided feedback.

This notebook shows how the [Model Alignment Python API](https://github.com/PAIR-code/model-alignment) can be used with Gemma and Gemini models to modify non-agentic prompts (e.g. ideation, extraction, summarization, etc. that don't make use of tools or external knowledge databases) based on  feedback through two different approaches described below.

Running this notebook requires a [GPU-backed Colab runtime][colab_accelerators], such as a T4, L4, or A100, which may require a paid [subscription][colab_subscriptions].

With model alignment, there are two roles that models take as part of the process, although these can be the same model in practice:

1. **Target model**: The model that the target prompt will be run with. The alignment process is meant to make the prompt work well with the target model, as determined by you.
  
  - This notebook uses the Gemma2 2B Instruction-Tuned model, loaded by Keras from weights hosted on Kaggle as the target model. Loading Gemma for this Colab requires:

    1. [Authenticating][kagglehub_auth] with a Kaggle account that has accepted the Gemma license agreement; and
    1. A [GPU-backed runtime][colab_accelerators], such as a T4, L4, or A100, which may require a paid [subscription][colab_subscriptions].

[colab_accelerators]: https://research.google.com/colaboratory/faq.html#gpu-availability
[colab_subscriptions]: https://colab.research.google.com/signup
[kagglehub_auth]: https://github.com/Kaggle/kagglehub#authenticate
[kaggle_gemma]: https://www.kaggle.com/models/keras/gemma
  
2. **Alignment model**: The model that will be used to perform the alignment steps. This model is used in the steps that perform alignment steps such as creating principles and updating the target prompt.

  - This notebook uses the Gemini Pro 2.5 model run through Google's Developer API as the alignment model. This requires you to provide an [API key](https://aistudio.google.com/app/apikey). The Gemini model is larger in size than the Gemma model, which makes it better at the alignment tasks shown in this notebook. A Gemma model larger than 2B would also suffice for the alignment of the prompts.

Alignment can be performed through two different methods, both of which are shown in this notebook:

1. Prompts are modified through guidlines (also called principles) which are curated by critiquing or rewriting of model responses.
2. Prompts are modified through direct feedback of a model response, without the intermediate step of creating principles.

The principles method has an added benefit of maintaining a list of principles by which the prompt should adhere, which can be helpful as both a resource to share and collaborate on, and as a checklist of properties that a model response can be judged agains. This does require extra alignment model processing and human involvement versus the direct feedback approach. Which approach to take is entirely up to you.

In addition to creating principles through critiquing, principles can also be created by providing a manually-written ideal response, or by giving kudos to a well-crafted model response. For more details, see the [documentation](https://github.com/PAIR-code/model-alignment?tab=readme-ov-file#generating-principles-from-feedback-and-applying-them) and [research paper](https://arxiv.org/abs/2310.15428).

# Simple Alignment Through Principles

This example shows aligning a simple prompt through the process of creating principles for the prompt to follow, created by providing feedback to responses generated by the model.

You first define a simple prompt to align.

You then run the prompt through the target Gemma model and see the output.

You can then provide a critique of the response in order to have the model alignment system create a generic principle that you want the model to adhere to. In this example, you ask the model alignment library to provide some possible critiques given this prompt and model response to help us brainstorm.

Once you provide the critique, a principle is created and printed out. You can repeat this process as many times as you want to update the full principles list for this model alignment task.

Once you are happy with the list of principles, you have the model alignment library update the prompt to better adhere to all of the principles you have created.

You then run the updated prompt on the target model and see the improved results.

```python
# @title Prompt to align
prompt = 'Give me a list of groceries to buy for making shrimp scampi'  # @param {type:"string"}
```

```python
# Create the model alignment instance, using Gemma to run the prompt and Gemini to do the alignment work.
single_run_prompt = single_run.AlignableSingleRun(gemma_model, gemini_model)
```

```python
# Run the initial prompt with Gemma
single_run_prompt.set_model_description(prompt)
output = single_run_prompt.send_input()
print(output.text)
```

```python
# Generate some possible critiques of this model output using the larger model.
critiques = single_run_prompt.generate_critiques()
print(critiques)
```

```python
# Make a specific critique of the prompt and response, creating a principle with the larger model.
# Then print the current principles list.
principles = single_run_prompt.critique_response(
    'I do not want to use any alcohol in my recipe'
)
print(principles)
```

```python
# Update the prompt based on the principles using the larger model and print it out.
single_run_prompt.update_model_description_from_principles()
print(single_run_prompt.get_model_description())
```

```python
# Get model output from Gemma now that the prompt is updated based on the new principle.
reply = single_run_prompt.send_input()
print(reply.text)
```

# Task
I will now design a Graph Neural Network (GNN) model architecture for detecting intrusions within the 3D cube.

Here's a proposed GNN architecture:

**1. Input Layer:**
*   **Node Features:** The input to our GNN will be the `feature_matrix` we've already generated, which contains 6 features for each node: network traffic (in/out), CPU load, memory load, file hash metric, and the local threat score.
*   **Graph Structure:** The adjacency matrix (or equivalent graph representation) derived from our `nx.grid_graph` will define the connections between nodes.

**2. GNN Layers (Core Message Passing):**
*   We'll employ **GraphSAGE (Graph Sample and Aggregate)** layers. GraphSAGE is well-suited for potentially large or "infinite" graph structures due to its inductive nature, allowing it to generalize to unseen nodes or evolving graph topologies. It learns aggregation functions instead of a fixed weight matrix for the entire graph.
*   **Layer 1 (Embedding Layer):**
    *   Input dimension: 6 (from our node features)
    *   Output dimension: `hidden_dim` (e.g., 64 or 128)
    *   Aggregator: `mean` (a common and effective aggregation function)
    *   Activation: `ReLU` (Rectified Linear Unit) for non-linearity, enabling the model to learn complex patterns.
    *   Regularization: `Dropout` (e.g., 0.5) to prevent overfitting.
*   **Layer 2 (Further Abstraction):**
    *   Input dimension: `hidden_dim`
    *   Output dimension: `hidden_dim` (can be different from Layer 1, but often kept the same for initial designs)
    *   Aggregator: `mean`
    *   Activation: `ReLU`
    *   Regularization: `Dropout` (e.g., 0.5)

**3. Output Layer (Task-Specific):**
*   **Task:** Node-level intrusion detection, specifically predicting a continuous threat score.
*   We'll use a `Dense` (fully connected) layer.
*   Input dimension: `hidden_dim` (from the last GNN layer)
*   Output dimension: 1 (to predict a single threat score for each node).
*   Activation: No activation (linear output) to allow for a continuous range of threat scores, or a `sigmoid` activation scaled to the expected range (e.g., 0-10) if we want to constrain the output. For simplicity, a linear output is a good starting point, letting the loss function guide its values.

**4. Loss Function:**
*   Given that our current `threat_score` is a continuous numerical value, we'll use **Mean Squared Error (MSE)** or **Mean Absolute Error (MAE)**. MSE penalizes larger errors more heavily, which might be desirable for critical intrusions.

**5. Optimizer:**
*   **Adam** (Adaptive Moment Estimation) is a widely used and effective optimizer for deep learning models.

**Summary of Architecture:**

`Input Node Features (6 dimensions) & Graph Adjacency`
      ↓
`GraphSAGE Layer (Output: hidden_dim, Activation: ReLU, Dropout: 0.5)`
      ↓
`GraphSAGE Layer (Output: hidden_dim, Activation: ReLU, Dropout: 0.5)`
      ↓
`Dense Layer (Output: 1 - Predicted Threat Score, Activation: Linear)`
      ↓
`Loss: Mean Squared Error`
`Optimizer: Adam`

This architecture allows the model to learn node embeddings by aggregating information from its neighbors, capturing local patterns of activity and potential anomalies. The sequential GraphSAGE layers enable the propagation of information through the 3D cube structure, helping to detect distributed or coordinated intrusions. The final Dense layer then translates these learned embeddings into a specific threat score for each core (node).

### Subtask:
Select appropriate Graph Neural Network layers (e.g., GCN, GraphSAGE, GAT) to perform message passing and learn node embeddings.


**Reasoning**:
The subtask requires defining the GNN layers according to the proposed architecture. This involves importing `tensorflow` and `keras`, setting `hidden_dim`, and then defining two conceptual GraphSAGE layers. Since native `GraphSAGE` layers aren't directly available in `tf.keras.layers`, I will implement them as custom Keras layers to correctly handle node feature transformation and aggregation using `Dense` layers and `ReLU` activation, with `Dropout` for regularization.



```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import keras as krs # Import keras again explicitly for clarity with Keras 3

# Define the hidden dimension as specified in the architecture
hidden_dim = 64 # Can be 64 or 128 as suggested

# Custom GraphSAGE Layer implementation
class GraphSAGE(layers.Layer):
    def __init__(self, output_dim, aggregator_type='mean', dropout_rate=0.5, activation='relu', **kwargs):
        super(GraphSAGE, self).__init__(**kwargs)
        self.output_dim = output_dim
        self.aggregator_type = aggregator_type
        self.dropout_rate = dropout_rate
        self.activation_fn = keras.activations.get(activation)
        self.dropout = layers.Dropout(dropout_rate)

    def build(self, input_shape):
        # input_shape[0] for node features, input_shape[1] for adjacency matrix
        input_dim = input_shape[0][-1] # Feature dimension of nodes

        # Dense layer for self-features
        self.self_dense = layers.Dense(self.output_dim, use_bias=True)
        # Dense layer for aggregated neighbor features
        self.neigh_dense = layers.Dense(self.output_dim, use_bias=False)

    def call(self, inputs):
        node_features, adjacency_matrix = inputs

        # 1. Aggregate neighbor features
        # Sum neighbor features according to adjacency matrix
        # Equivalent to sum(A_ij * h_j) for all j neighbors of i
        if self.aggregator_type == 'mean':
            # Normalize adjacency matrix to get mean
            degrees = tf.reduce_sum(adjacency_matrix, axis=1, keepdims=True)
            # Avoid division by zero for isolated nodes
            degrees = tf.where(tf.equal(degrees, 0), tf.ones_like(degrees), degrees)
            norm_adjacency_matrix = adjacency_matrix / degrees
            aggregated_features = tf.matmul(norm_adjacency_matrix, node_features)
        elif self.aggregator_type == 'sum':
            aggregated_features = tf.matmul(adjacency_matrix, node_features)
        else:
            raise ValueError(f"Aggregator type {self.aggregator_type} not supported")

        # 2. Transform self-features and aggregated features
        self_transformed = self.self_dense(node_features)
        neigh_transformed = self.neigh_dense(aggregated_features)

        # 3. Concatenate and apply activation (or sum and apply activation)
        # For GraphSAGE, it's typically a concatenation or element-wise sum followed by activation
        # Here, we'll sum them for simplicity as a common practice, then activate.
        # A more 'classic' GraphSAGE would concatenate and then pass through a Dense layer.
        combined_features = self_transformed + neigh_transformed
        output = self.activation_fn(combined_features)
        output = self.dropout(output)
        return output

    def get_config(self):
        config = super(GraphSAGE, self).get_config()
        config.update({
            'output_dim': self.output_dim,
            'aggregator_type': self.aggregator_type,
            'dropout_rate': self.dropout_rate,
            'activation': keras.activations.serialize(self.activation_fn)
        })
        return config

print(f"Defined hidden_dim: {hidden_dim}")
print("GraphSAGE custom layer defined.")
```

The subtask instructs to instantiate the two GraphSAGE layers and the final output layer as defined in the proposed architecture. This will complete the core GNN layers and the output layer selection.
